```python
import pandas as pd

# Создание объекта Series из списка с использованием параметров pd.Series()
countries = pd.Series(
    data = ['Англия', 'Канада', 'США', 'Россия', 'Украина', 'Беларусь', 'Казахстан'],
    index = ['UK', 'CA', 'US', 'RU', 'UA', 'BY', 'KZ'],
    name = 'countries'
)

# display используется для вывода табличных данных по аналогии с функцией print
display(countries)

# Если оставить параметр index пустым, то метки (индексы) будут присвоены автоматически в виде
# порядковых номеров элементов
```

```python
import pandas as pd

# Создание объекта Series из словаря. Метки (индексы) — ключи словаря
# значения Series — значения словаря

countries = pd.Series({
    'UK': 'Англия',
    'CA': 'Канада',
    'US': 'США',
    'RU': 'Россия',
    'UA': 'Украина',
    'BY': 'Беларусь',
    'KZ': 'Казахстан'},
    name = 'countries'
)
display(countries)
```

Доступ к данным в Series осуществляется через loc или iloc.

.loc вызывается с квадратными скобками, в которые передаются метки. В него можно передать как один индекс, так
и список, чтобы получить сразу несколько элементов.

```python
# При обращении по одному индексу возвращается строка
print(countries.loc['US'])

# При обращении по нескольким элементам, возвращается объект Series
print(countries.loc[['US', 'RU', 'UK']])
```

.iloc также вызывается с квадратными скобками и принимает на вход порядковые номера элементов Series
(нумерация начинается с 0). В него можно передавать как один индекс, так и диапазон чисел

```python
print(countries.iloc[6])

# Срез из исходной Series с первого по третий элемент (конечное значение не включается в результат)
print(countries.iloc[1:4])
```

loc и iloc можно опустить и обращаться к элементам Series напрямую по индексам, но при обращении к DataFrame они необходимы

```python
import pandas as pd

def create_medications(names, counts):
    # Создаем серию с индексами названиями лекарств, и значениями количеством
    medications = pd.Series(index=names, data=counts)
    return medications
def get_percent(medications, name):
    # Находим долю конкретного лекарства от общего количества (в процентах)
    return medications.loc[name]/sum(medications) * 100
```
