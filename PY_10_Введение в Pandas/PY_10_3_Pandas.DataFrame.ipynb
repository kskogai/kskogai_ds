Для работы с табличными данными в Pandas существует объект DataFrame.

DataFrame является двумерной структурой и представляется в виде таблицы, в которой есть строки и столбцы: столбцами в DataFrame выступают объекты Series, а строки формируются из их элементов. Также в DataFrame есть метки (индексы), которые соответствуют каждой строке таблицы.

DataFrame и таблица — синонимы.
Столбец таблицы и признак — синонимы.

DataFrame создается с помощью функции pd.DataFrame()

Самый простой способ создания DataFrame — из словаря, ключами которого являются имена столбцов таблицы, а значениями — списки, в которых хранится содержимое этих столбцов.

```python
import pandas as pd

countries_df = pd.DataFrame({
    'country': ['Англия', 'Канада', 'США', 'Россия', 'Украина', 'Беларусь', 'Казахстан'],
    'population': [56.29, 38.05, 322.28, 146.24, 45.5, 9.5, 17.04],
    'area': [133396, 9984670, 9826630, 171125191, 603628, 207600, 2724902]
})

display(countries_df)
```

Так как не были заданы метки, они сгенерировались автоматически (по номерам, начиная с 0). Метки можно задать вручную.

```python
countries_df.index = ['UK', 'CA', 'US', 'RU', 'UA', 'BY', 'KZ']
display(countries_df)
```

Также DataFrame можно создать из вложенного списка, внутренние списки которого будут являться строками новой таблицы.

```python
countries_df = pd.DataFrame(
    data = [
        ['Англия', 56.29, 133396],
        ['Канада', 38.05, 9984670],
        ['США', 322.28, 9826630],
        ['Россия', 146.24, 17125191],
        ['Украина', 45.5, 603628],
        ['Беларусь', 9.5, 207600],
        ['Казахстан', 17.04, 2724902]
    ],
    columns= ['country', 'population', 'area'],
    index= ['UK', 'CA', 'US', 'RU', 'UA', 'BY', 'KZ']
)
display(countries_df)
```

Так как DataFrame — двумерная структура данных, существует два возможных направления движения в ней: вдоль строк и вдоль столбцов.

Например, мы можем удалять строки и столбцы таблицы, вычислять среднее значение как по столбцам, так и по строкам.

При работе с Pandas важно уметь указывать направление работы метода, который используется. Для этого вводится понятие axis (ось, координата). Движение по строкам в таблице обозначается как axis с индексом 0, а движение по столбцам — axis с индексом 1.

Данный параметр заложен во все методы, которые могут работать в двух направлениях, и по умолчанию в большинстве из них axis=0, то есть они выполняют операции со строками, если не задавать axis вручную.

```python
# Среднее по таблице при axis=0, то есть среднее по строкам в каждом столбце
display(countries_df.mean(axis=0, numeric_only=True))

# Среднее по таблице при axis=1, то есть среднее по столбцам в каждой строке
display(countries_df.mean(axis=1, numeric_only=True))
```

Доступ к столбцу можно получить разными способами.

Способ 1: обратиться к DataFrame по имени столбца через точку. Однако использование такого способа возможно только когда имя столбца указано без пробелов.

```python
display(countries_df.population)
```

Способ 2: обратиться к DataFrame по индексу и указать имя столбца. 

```python
display(countries_df['population'])
```

При обращении к столбцу DataFrame мы получаем объект Series с именем, соответствующим имени столбца.

```python
type(countries_df.population)
```

Чтобы получить доступ к ячейкам таблицы, используются loc и iloc.

При этом, в соответствии с механизмом работы axis, при обращении к DataFrame по индексам с помощью loc(iloc) первым индексом указывается индекс (порядковый номер), соответствующий строкам, а вторым - имя (порядковый номер) столбца.

```python
# Получим площадь Великобритании
countries_df.loc['UK', 'area']
```

```python
# Получим население и площадь России
countries_df.loc['RU', ['population', 'area']]
```

```python
# Сделаем вырезку из таблицы и получим информацию о населении и площади Украины, Беларуси и Казахстана
countries_df.loc[['UA', 'BY', 'KZ'], ['population', 'area']]
```

```python
# или
countries_df.iloc[4:8, 1:3]
```

```python
import pandas as pd

# Создаем датафрейм, в котором индексы — годы, а столбцы — доход и расходы
def create_companyDF(income, expenses, years):
    df = pd.DataFrame({
        'Income': income,
        'Expenses': expenses
    },
    index = years
    )
    return df

# Создаем функцию, которая сначала проверяет наличие индекса в таблице,
# и если индекс есть, то происходит вычисление разницы между доходами и расходами
# за нужный год
# Если такого индекса нет, то возвращается None
def get_profit(df, year):
    if year in df.index:
        profit = df.loc[year, 'Income'] - df.loc[year, 'Expenses']
    else:
        profit = None
    return profit
```
