```python
import pandas as pd

dates = pd.read_csv('data/movies_data/dates.csv')
movies = pd.read_csv('data/movies_data/movies.csv')
ratings1 = pd.read_csv('data/movies_data/ratings1.csv')
ratings2 = pd.read_csv('data/movies_data/ratings2.csv')
```

Первым делом необходимо склеить таблицы ratings1 и ratings2 по строкам.

Для этого воспользуемся встроенной функцией Pandas concat(), которая позволяет конкатенировать (склеивать) таблицы как по строкам, так и по столбцам.

Основные параметры функции concat():

- objs — список объектов DataFrame ([df1, df2, ...]), которые должны быть сконкатенированы;

- axis — ось определяет направление конкатенации: 0 — конкатенация по строкам (по умолчанию), 1 — конкатенация по столбцам;

- join — либо inner (пересечение), либо outer (объединение);

- ignore_index — по умолчанию установлено значение False, которое позволяет значениям индекса оставаться такими, какими они были в исходных данных. Если установлено значение True, параметр будет игнорировать исходные значения и повторно назначать значения индекса в последовательном порядке.

Для корректной конкатенации по строкам объединяемые таблицы должны иметь одинаковую структуру — одинаковое количество и имена столбцов.

ratings1 и ratings2 имеют одинаковую структуру столбцов, поэтому их можно склеить по строкам, передав их списком в функцию concat(). Параметр axis по умолчанию равен 0, так как объединение происходит по строкам.

ПРИМЕЧАНИЕ:

concat является функцией библиотеки, а не методом датафрейма, поэтому ее вызов осуществляется через pd.concat(...).

```python
ratings = pd.concat([ratings1, ratings2])
display(ratings)
```

В результате мы увеличили первую таблицу, добавив снизу строки второй таблицы.

НО:

Если обратить внимание на индексы последних строк, то видно, что их нумерация не совпадает с количеством строк. Это может привести к некорректному объединению таблиц по ключевым столбцам на следующем этапе решения задачи.

Это связано с тем, что по умолчанию concat сохраняет первоначальные индексы объединяемых таблиц, а обе наши таблицы индексировались, начиная с 0. Чтобы создать новые индексы, нужно выставить параметр ignore_index на True:

```python
ratings = pd.concat(
    [ratings1, ratings2],
    ignore_index=True
)

display(ratings)
```

Далее узнаем количество строк в таблицах ratings и dates, потому что их тоже необходимо будет склеить между собой:

```python
print(f'Число строк в таблице ratings: {ratings.shape[0]}')
print(f'Число строк в таблице dates: {dates.shape[0]}')
print(ratings.shape[0] == dates.shape[0])
```

Почему размерность таблиц различается?

При выгрузке данных информация об оценках какого-то пользователя попала в обе таблицы ratings1 и ratings2. В результате конкатенации случилось дублирование строк. В данном случае их можно легко найти выводом последней строки таблицы ratings1 и первой строки таблицы ratings2:

```python
display(ratings1.tail(1))
display(ratings2.head(1))
```

Чтобы очистить таблицу от дублей, можно воспользоваться методом DataFrame drop_duplicates(), который удаляет повторяющиеся строки в таблице. Не забываем обновить индексы после удаления дублей внутри метода drop_duplicates (ignore_index=True):

```python
ratings = ratings.drop_duplicates(ignore_index=True)
print(f'Число строк в таблице ratings: {ratings.shape[0]}')
```

Теперь можно добавить к таблице с рейтингами даты их выставления. Для этого нужно конкатенировать таблицы ratings и dates по столбцам:

```python
ratings_dates = pd.concat([ratings, dates], axis=1)
display(ratings_dates.tail(7))
```
