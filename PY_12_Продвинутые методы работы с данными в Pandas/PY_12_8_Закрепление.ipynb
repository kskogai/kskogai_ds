```python
import pandas as pd

movies_df = pd.read_csv('data/ratings_movies.csv')
display(movies_df)
```

```python
#библиотека для регулярных выражений
import re 
def get_year_release(arg):
    #находим все слова по шаблону "(DDDD)"
    candidates = re.findall(r'\(\d{4}\)', arg) 
    # проверяем число вхождений
    if len(candidates) > 0:
        #если число вхождений больше 0,
	#очищаем строку от знаков "(" и ")"
        year = candidates[0].replace('(', '')
        year = year.replace(')', '')
        return int(year)
    else:
        #если год не указан, возвращаем None
        return None
```

```python
movies_df['year_release'] = movies_df['title'].apply(get_year_release).astype('Int32')
display(movies_df)
```

```python
movies_df['year_release'].info()
```

```python
movies_df[movies_df['year_release'] == 1999].groupby('title')['rating'].agg(['mean', 'min']).sort_values(by=['mean', 'min'], ascending=[True, True])
```

```python
movies_df[movies_df['year_release'] == 2010].groupby('genres')['rating'].agg(['mean', 'min']).sort_values(by=['mean', 'min'], ascending=[True, True])
```

```python
movies_df.groupby('userId')['genres'].agg(
    ['nunique', set]
).sort_values(by='nunique', ascending=False)
```

```python
movies_df.groupby('userId')['rating'].agg(
    ['nunique', 'mean']
).sort_values(by='mean', ascending=False)
```

```python
mask = movies_df['year_release'] == 2018
grouped = movies_df[mask].groupby('genres')['rating'].agg(
    ['mean', 'count']
)
grouped[grouped['count'] > 10].sort_values(by='mean', ascending=False)
```

```python
movies_df['date'] = pd.to_datetime(movies_df['date'])
movies_df['year_rating'] = movies_df['date'].dt.year

pivot = movies_df.pivot_table(
    values='rating',
    index='year_rating',
    columns='genres',
    aggfunc='mean'
)

pd.set_option('display.max_rows', None)
display(pivot.loc[2018].sort_values(ascending=False))
```

```python
orders_df = pd.read_csv('data/orders.csv', sep=';')
products_df = pd.read_csv('data/products.csv', sep=';')

orders_products = orders_df.merge(
    products_df,
    left_on='ID товара',
    right_on='Product_ID',
    how='left'
)

orders_products[orders_products['Name'].isna()]

```

```python
orders_products['Profit'] = orders_products['Количество'] * orders_products['Price']
mask = orders_products['Оплачен'] == 'Да'
grouped = orders_products[mask].groupby('ID Покупателя')['Profit'].sum().sort_values(ascending=False)
display(grouped)
```
