```python
bike_df.drop(['start station id', 'end station id'], axis=1, inplace=True)
display(bike_df)
```

```python
bike_df['age'] = 2018-bike_df['birth year']
bike_df.drop(['birth year'], axis=1, inplace=True)
display(bike_df[bike_df['age'] > 60])
```

```python
bike_df['starttime'] = pd.to_datetime(bike_df['starttime'])
bike_df['stoptime'] = pd.to_datetime(bike_df['stoptime'])
bike_df['trip duration'] = bike_df['stoptime']-bike_df['starttime']
display(bike_df.loc[3, 'trip duration'])
```

```python
weekday = bike_df['starttime'].dt.dayofweek
bike_df['weekend'] = weekday.apply(lambda x: 1 if x==5 or x==6 else 0)
bike_df['weekend'].sum()
```

```python
def get_time_of_day(time):
    if 0 <= time <= 6:
        return 'night'
    if 6 < time <= 12:
        return 'morning'
    if 12 < time <= 18:
        return 'day'
    return 'evening'

hour = bike_df['starttime'].dt.hour
bike_df['time of day'] = hour.apply(get_time_of_day)

day_rides = bike_df[bike_df['time of day'] == 'day'].shape[0]
night_rides = bike_df[bike_df['time of day'] == 'night'].shape[0]
display(round(day_rides/night_rides))
```
