Формат DATETIME

Многие выгрузки из систем и баз данных имеют свой служебный формат:

- 2018-11-09 15:45:21;
- 11/09/2018 3:45:20 PM;
- 2018-11-09T15:45:21.2984

Для всех этих случаев необходимо задавать формат распознавания дат и уметь сравнивать их между собой. Для этого был создан единый способ обозначения даты и времени.

Таким форматом в Pandas является формат datetime, который записывается как YYYY-MM-DD HH: MM: SS, то есть составляющие времени указываются в порядке: год, месяц, день, час, минута, секунда.

В наших данных дата записана в виде DD/MM/YYYY

```python
display(melb_df['Date'])
```

Для преобразования столбцов с датами, записанными в распространенных форматах, в формат datetime можно воспользоваться функцией pandas.to_datetime(). В нашем случае в функции нужно указать параметр dayfirst=True, который будет обозначать, что в первоначальном признаке первым идет день. Преобразуем столбец Date в формат datetime, передав в него функцию

```python
melb_df['Date'] = pd.to_datetime(melb_df['Date'], dayfirst=True)
display(melb_df['Date'])
```

В результате мы переопределяем признак Date в формат datetime. При этом так как в изначальном варианте время не было указано, то и после преобразования оно опускается.

Стоит обратить внимание, что изменился тип данных столбца Date, теперь его тип — datetime64.

Тип данных datetime позволяет с помощью специального аксессора dt выделять составляющие времени из каждого элемента столбца, такие как:

- date — дата;

- year, month, day — год, месяц, день;

- time — время;

- hour, minute, second — час, минута, секунда;

- dayofweek — номер дня недели, от 0 до 6, где 0 — понедельник, а 6 — воскресенье;

- day_name — название дня недели;

- dayofyear — порядковый день года;

- quarter — квартал (интервал в 3 месяца).

Аксессор — это атрибут столбца, хранящий переменные, которые были строковым представлением времени, а затем были изменены в помощью pd.to_datetime().

К аксессору нельзя обратиться, если столбец не приведен к типу datetime.

Обратившись по атрибуту dt.year в столбце Date, можно достать год продажи и понять, за какой интервал времени (в годах) представлены наши данные, а также на какой год приходится наибольшее число продаж.

```python
years_sold = melb_df['Date'].dt.year
print(years_sold)
print('Min year sold:', years_sold.min())
print('Max year sold:', years_sold.max())
print('Mode year sold:', years_sold.mode()[0])
```

В результате обращения к атрибуту datetime melb_df['Date'].dt.year мы получаем объект Series, в котором в качестве значений выступают годы продажи объектов недвижимости. Мы можем занести результат в переменную years_sold и далее работать с ней как с обычным столбцом серии — вычислять минимум, максимум и модальное значение (моду).

Из результатов видно, что данные представлены за интервал с 2016 по 2017 год, и наибольшее число объектов было продано в 2017 году.

Так как модальных значений в столбце может быть несколько, метод mode() возвращает объект Series, даже если мода в данных только одна. Чтобы сохранить стилистику вывода информации о годе продажи и выводить только число, а не Series, мы обращаемся к результату работы метода mode() по индексу 0.

Теперь вычислим, на какие месяцы приходится пик продаж объектов недвижимости. Для этого выделим атрибут dt.month и на этот раз занесем результат в столбец MonthSale, а затем найдем относительную частоту продаж для каждого месяца от общего числа продаж — для этого используем value_counts() с параматером normalize (вывод в долях).

```python
melb_df['MonthSale'] = melb_df['Date'].dt.month
melb_df['MonthSale'].value_counts(normalize=True)
```

Из результатов становится ясно, что наибольшее количество продаж недвижимости приходится на май, июль и сентябрь (5, 7, 9 месяцы). Месяцами застоя являются октябрь, февраль и январь (10, 2, 1).

Работа с интервалами

Часто бывает ситуация, что необходимо вычислять интервалы между двумя временными промежутками. Например, можно вычислить, сколько дней прошло с 1 января 2016 года до момента продажи объекта. Для этого можно просто найти разницу между датами продаж и заявленной датой, представленной в формате datetime

```python
delta_days = melb_df['Date'] - pd.to_datetime('2016-01-01')
display(delta_days)
```

В результате получаем серию, элементами которой является количество дней, которое прошло с 1 января 2016 года. Данные такого формата относятся к типу timedelta.

Чтобы превратить количество дней из формата интервала в формат целого числа дней, можно воспользоваться аксессором dt для формата timedelta и извлечь из него атрибут days:

```python
display(delta_days.dt.days)
```

Создадим признак возраста объекта недвижимости в годах на момент продажи. Для этого выделим из столбца с датой продажи год и вычтем из него год постройки здания. Результат оформим в виде столбца AgeBuilding

```python
melb_df['AgeBuilding'] = melb_df['Date'].dt.year - melb_df['YearBuilt']
display(melb_df['AgeBuilding'].astype('int16'))
```

Столбец AgeBuilding дублирует информацию из столбца YearBuilt. Поэтому есть смысл оставить возраст здания, так как он более наглядный, а год постройки удалить из таблицы.

```python
melb_df = melb_df.drop('YearBuilt', axis=1)
```

Создайте в таблице melb_df признак WeekdaySale (день недели). Найдите, сколько объектов недвижимости было продано в выходные (суббота и воскресенье), результат занесите в переменную weekend_count

```python
melb_df['WeekdaySale'] = melb_df['Date'].dt.dayofweek
weekend_count = melb_df[(melb_df['WeekdaySale'] == 5) | (melb_df['WeekdaySale'] == 6)].shape[0]
display(weekend_count)
```
